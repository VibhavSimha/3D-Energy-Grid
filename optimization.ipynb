```python
# Cost-based values
generation_cost_values = {
    "generation solar": 2,
    "generation wind onshore": 1,
    "generation hydro run-of-river and poundage": 3,
    "generation hydro water reservoir": 4,
    "generation hydro pumped storage consumption": 5,
    "generation other renewable": 6,
    "generation biomass": 7,
    "generation fossil hard coal": 8,
    "generation fossil brown coal/lignite": 9,
    "generation fossil oil": 10,
    "generation fossil gas": 6,
    "generation waste": 7,
    "generation nuclear": 8,
    "generation other": 0
}

# Data: average load, environmental impact, and damage
generation_data = {
    "generation solar": {"avg_load": 1432.818546087155, "impact_value": 5, "damage_value": 0},
    "generation wind onshore": {"avg_load": 5464.980450034223, "impact_value": 4, "damage_value": 0},
    "generation hydro run-of-river and poundage": {"avg_load": 972.2019022359115, "impact_value": 3, "damage_value": 0},
    "generation hydro water reservoir": {"avg_load": 2605.5341233173626, "impact_value": 2, "damage_value": 0},
    "generation hydro pumped storage consumption": {"avg_load": 475.5827059091946, "impact_value": 2, "damage_value": 0},
    "generation other renewable": {"avg_load": 85.63432580424367, "impact_value": 1, "damage_value": 0},
    "generation biomass": {"avg_load": 383.53134268765683, "impact_value": 1, "damage_value": 0},
    "generation fossil hard coal": {"avg_load": 4256.531271389459, "impact_value": -6, "damage_value": -6},
    "generation fossil brown coal/lignite": {"avg_load": 448.0945699292722, "impact_value": -5, "damage_value": -5},
    "generation fossil oil": {"avg_load": 298.34241672370524, "impact_value": -4, "damage_value": -4},
    "generation fossil gas": {"avg_load": 5622.7006473876345, "impact_value": -3, "damage_value": -3},
    "generation waste": {"avg_load": 269.418691535478, "impact_value": -2, "damage_value": -2},
    "generation nuclear": {"avg_load": 6263.483430298882, "impact_value": -1, "damage_value": -1},
    "generation other": {"avg_load": 60.22602954597308, "impact_value": 0, "damage_value": 0}
}

from typing import Dict, List, Tuple, Optional

# Type definitions for clarity
EnergyData = Dict[str, Dict[str, float]]
CostValues = Dict[str, int]

def normalize_weight(weight: float) -> float:
    """Normalize weight to ensure it is a valid value between 0 and 1."""
    if weight < 0 or weight > 1:
        raise ValueError("Weight must be between 0 and 1.")
    return weight

def validate_load(load: float, total_available: float) -> None:
    """Validate if the requested load can be met with available capacity"""
    if load <= 0:
        raise ValueError("Load must be positive")
    if load > total_available:
        raise ValueError(f"Requested load ({load}) exceeds total available capacity ({total_available:.2f})")

def calculate_total_available_load(generation_data: EnergyData) -> float:
    """Calculate total available generation capacity"""
    return sum(data['avg_load'] for data in generation_data.values())

def find_optimal_energy_combo(
    user_load: float,
    generation_data: EnergyData,
    generation_cost_values: CostValues,
    cost_weight: float = 1.0,  # Only cost-based optimization
    min_source_contribution: float = 0.0  # Minimum load percentage for a source to be included
) -> Tuple[List[Tuple[str, float, int, int, int]], float]:
    """
    Find optimal combination of energy sources to meet required load based on cost optimization.

    Args:
        user_load: Required load to meet
        generation_data: Dictionary containing generation data for each source
        generation_cost_values: Dictionary containing cost values for each source
        cost_weight: Weight given to cost in optimization (0-1)
        min_source_contribution: Minimum percentage of total load for a source to be included

    Returns:
        Tuple containing list of selected sources and total generated load
    """
    # Validate inputs
    cost_weight = normalize_weight(cost_weight)
    total_available = calculate_total_available_load(generation_data)
    validate_load(user_load, total_available)

    def cost_score(source: str) -> float:
        """Calculate cost score for a source considering cost only"""
        return generation_cost_values[source] * cost_weight

    # Sort sources by cost score (lowest cost first)
    sorted_sources = sorted(generation_data.items(), key=lambda x: cost_score(x[0]))

    # Initialize result variables
    total_load = 0
    selected_sources = []
    min_load_threshold = user_load * min_source_contribution

    # Select sources to meet load requirement based on cost
    for source, data in sorted_sources:
        if total_load >= user_load:
            break

        # Only include sources that contribute meaningfully to the total load
        if data['avg_load'] >= min_load_threshold:
            total_load += data['avg_load']
            selected_sources.append((source, data['avg_load'], generation_cost_values[source], data['impact_value'], data['damage_value']))

    return selected_sources, total_load

def format_output(selected_sources: List[Tuple[str, float, int, int, int]],
                 total_generated: float,
                 user_load: float) -> str:
    """Format the output results into a readable string"""
    output = [f"\nOptimal Energy Generation Plan for {user_load:.2f} units (Cost-based):"]
    output.append("\nSelected Sources:")

    total_cost = 0

    for source, load, cost, impact, damage in selected_sources:
        percentage = (load / total_generated) * 100
        total_cost += cost

        output.append(
            f"\n{source}:"
            f"\n  Load: {load:.2f} units ({percentage:.1f}%)"
            f"\n  Cost Value: {cost}"
            f"\n  Environmental Impact: {impact}"
            f"\n  Damage Value: {damage}"
        )

    output.append(f"\nTotal Generation: {total_generated:.2f} units")
    output.append(f"Average Cost Value: {total_cost/len(selected_sources):.2f}")

    return "\n".join(output)

def main():
    try:
        # Get user input with validation
        while True:
            try:
                user_input_load = float(input("Enter the required load: "))
                if user_input_load > 0:
                    break
                print("Please enter a positive number.")
            except ValueError:
                print("Please enter a valid number.")

        # Find optimal combination based on cost
        selected_sources, total_generated = find_optimal_energy_combo(
            user_input_load,
            generation_data,
            generation_cost_values,
            cost_weight=1.0,  # Pure cost-based optimization
            min_source_contribution=0.05  # 5% minimum contribution
        )

        # Display results
        print(format_output(selected_sources, total_generated, user_input_load))

    except ValueError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

if __name__ == "__main__":
    main()
```
